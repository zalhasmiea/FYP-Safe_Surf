```python
import matplotlib.pyplot as plt

# Data
labels = ['Legitimate', 'Phishing']
sizes = [80, 20]
colors = ['#c6f6d5', '#feb2b2']

# Create a new style: donut-style pie chart with enhanced design
fig, ax = plt.subplots()

# Create a donut chart
wedges, texts, autotexts = ax.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    startangle=140,
    colors=colors,
    wedgeprops={'linewidth': 1.5, 'edgecolor': 'white'},
    textprops={'color': '#2d3748', 'fontsize': 12},
    pctdistance=0.75
)

# Draw center circle for donut effect
centre_circle = plt.Circle((0, 0), 0.50, fc='white')
fig.gca().add_artist(centre_circle)

# Set title and layout
ax.set_title('URL Classification Breakdown', color='#2d3748', fontsize=16, pad=20)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
months = ['April', 'May', 'June']
total_scans = [5000, 5400, 5587]
malicious_counts = [900, 1050, 1247]
legitimate_counts = [t - m for t, m in zip(total_scans, malicious_counts)]

# Plot
fig, ax = plt.subplots(figsize=(8, 6), facecolor='none')

# Bars
ax.bar(months, legitimate_counts, label='Legitimate', color='#c6f6d5')
ax.bar(months, malicious_counts, bottom=legitimate_counts, label='Phishing', color='#feb2b2')

# Remove axis spines (the border lines)
for spine in ax.spines.values():
    spine.set_visible(False)

# Styling
# ax.set_ylabel('URLs Scanned')  # Removed side title as before
ax.legend()
ax.grid(False)
ax.set_facecolor('none')  # Transparent background
fig.patch.set_alpha(0)    # Transparent figure background
ax.set_title('')          # No title

plt.tight_layout()
plt.show()

```
